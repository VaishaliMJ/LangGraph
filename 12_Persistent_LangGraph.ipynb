```python
from langgraph.graph import START,END,StateGraph
from typing import TypedDict
from langchain_ollama import ChatOllama
from langgraph.checkpoint.memory import InMemorySaver
```

```python
llm=ChatOllama(model="llama3")
```

```python
class JokeState(TypedDict):
    topic   :   str
    joke    :   str
    explaination    :   str
    
```

```python
def generateJoke(state : JokeState):
    prompt  =   f"Generate the joke on the topic : {state["topic"]}"
    response    =   llm.invoke(prompt).content
    return {'joke'  :   response}
```

```python
def generateExplaination(state  :   JokeState):
    prompt  =   f"Write an explaination for the joke    :   {state['joke']}"
    response    =   llm.invoke(prompt).content
    return {'explaination'  :   response}
```

```python
graph = StateGraph(JokeState)

graph.add_node("generateJoke",generateJoke)
graph.add_node("generateExplaination",generateExplaination)

graph.add_edge(START,"generateJoke")
graph.add_edge("generateJoke","generateExplaination")
graph.add_edge("generateExplaination",END)


```

```python
checkpointer    =   InMemorySaver()

workflow    =   graph.compile(checkpointer=checkpointer)

config1 =   {"configurable" :   {"thread_id" :   "1"}} 

workflow.invoke({"topic"    :   "pizza"},config=config1)   
```

```python
workflow.get_state(config1)
```

```python
config2 =   {"configurable" :   {"thread_id" :   "2"}} 

workflow.invoke({"topic"    :   "pasta"},config=config2)   
```
