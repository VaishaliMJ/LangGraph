```python
from langchain_ollama import ChatOllama
from langgraph.graph import START,END,StateGraph
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model=ChatOllama(model="llama3", temperature=0)
```

```python
class LLMState(TypedDict):
    question:str
    answer:str
```

```python
def llmQA(state:LLMState)->LLMState:
    question=state['question']
    
    prompt=f"Answer from the following question {question}"
    
    
    answer=model.invoke(prompt).content
    
    state['answer']=answer
    
    return state
```

```python
graph=StateGraph(LLMState)


graph.add_node("llmQA",llmQA)

graph.add_edge(START,"llmQA")

graph.add_edge("llmQA",END)


workflow=graph.compile()
```

```python
initialState={"question" : "How many planets are there in our Solar system?"}

finalState=workflow.invoke(initialState)


print(finalState)
```
