```python
from langchain_ollama import ChatOllama
from langgraph.graph import START,END,StateGraph
from typing import TypedDict
```

```python
class BatsmanScore(TypedDict):
    runs : int
    balls : int
    fours : int
    sixes : int
    
    strikeRate : float
    ballsPerBoundary : float
    boundaryPercentage : float
    summary : str
    
```

```python
def calculateStrikeRate(state:BatsmanScore)->BatsmanScore:
    
    strikeRate=(state['runs']/state['balls'])*100
    
    return {'strikeRate':strikeRate}
```

```python
def calculateBallsPerBoundary(state:BatsmanScore)->BatsmanScore:
    
    ballsPerBoundary=state['balls']/(state['fours']+state['sixes'])
    
    return {'ballsPerBoundary':ballsPerBoundary}
```

```python
def calculateBoundaryPercentage(state:BatsmanScore)->BatsmanScore:
    
    boundaryPercentage=(((state['fours']*4) + (state['sixes']*6)) / state['runs'])*100
    
    return {'boundaryPercentage':boundaryPercentage}
```

```python
def summary(state:BatsmanScore):
    summary=f"""
    Strike rate - {state['strikeRate']}\n
    Balls Per Boundary - {state['ballsPerBoundary']}\n
    Boundary Percentage - {state['boundaryPercentage']} \n
    """
    
    return {'summary':summary}
```

```python
graph=StateGraph(BatsmanScore)

graph.add_node('calculateStrikeRate',calculateStrikeRate)
graph.add_node('calculateBallsPerBoundary',calculateBallsPerBoundary)
graph.add_node('calculateBoundaryPercentage',calculateBoundaryPercentage)
graph.add_node('summary',summary)

graph.add_edge(START,'calculateStrikeRate')
graph.add_edge(START,'calculateBallsPerBoundary')
graph.add_edge(START,'calculateBoundaryPercentage')

graph.add_edge('calculateStrikeRate','summary')
graph.add_edge('calculateBallsPerBoundary','summary')
graph.add_edge('calculateBoundaryPercentage','summary')

graph.add_edge('summary',END)

workflow = graph.compile()

workflow
```

```python
initialState={
    'runs' : 100,
    'balls' :50,
    'fours' : 6,
    'sixes' :4
}

workflow.invoke(initialState)
```
